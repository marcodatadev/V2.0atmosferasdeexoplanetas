# Actividad 4: Ejemplos de atmósferas de planetarias

Ahora que tenemos una idea de cómo funcionan los <span title="Un espectro es un gráfico del tamaño del planeta versus la longitud de onda"><font color='green'>espectros de transmisión</font></span>, consideremos diferentes tipos de modelos. Los tamaños atmosféricos se han hecho más grandes que la realidad para que sean más fáciles de ver.

#### Una atmósfera de vapor de agua


El siguiente modelo atmosférico contiene vapor de agua. Las moléculas de agua vibrarán y rotarán en algunas <span title="En una onda periódica la longitud de onda es la distancia física entre dos puntos a partir de los cuales la onda se repite"><font color='green'>longitudes de onda</font></span> mejor que en otras, por lo que el planeta se verá más grande en esas longitudes de onda cercanas a los 2,6 <span title="Un micrón es una unidad de longitud que es una millonésima parte de un metro. El cabello humano tiene unas 75 micras de diámetro."><font color='green'>micrones</font></span>.

<div class='alert alert-info'>
    <font size='3'><b>Inspecciona el espectro de abajo.</b></font>
</div>


```python
lightcurve_sliderES.example_spectra(atmospheres=['H2O'])
```

#### Una atmósfera de metano

El siguiente modelo atmosférico contiene metano. Al igual que el agua, las moléculas de metano vibrarán y rotarán en algunas <span title="En una onda periódica la longitud de onda es la distancia física entre dos puntos a partir de los cuales la onda se repite"><font color='green'>longitudes de onda</font></span> mejor que a otras. Sin embargo, el metano tiene una configuración diferente de átomos, por lo que el planeta parece más grande, cerca de 3,4 <span title="Un micrón es una unidad de longitud que es una millonésima parte de un metro. El cabello humano tiene unas 75 micras de diámetro."><font color='green'>micrones</font></span>.

<div class='alert alert-info'>
    <font size='3'><b>Inspecciona el espectro de abajo.</b></font>
</div>



```python
lightcurve_sliderES.example_spectra(atmospheres=['CH4'])
```

#### Una atmósfera de dióxido de carbono

El dióxido de carbono es otra configuración de moléculas con dos átomos de oxígeno en lados opuestos del carbono. La simetría de la molécula significa que solo hay unas pocas formas de hacer vibrar el dióxido de carbono. Este planeta se verá más grande en 2,8 <span title="Un micrón es una unidad de longitud que es una millonésima parte de un metro. El cabello humano tiene unas 75 micras de diámetro."><font color='green'>micrones</font></span> y 4,4 <span title="Un micrón es una unidad de longitud que es una millonésima parte de un metro. El cabello humano tiene unas 75 micras de diámetro."><font color='green'>micrones</font></span> pero es más pequeño en la mayoría de las otras  <span title="En una onda periódica la longitud de onda es la distancia física entre dos puntos a partir de los cuales la onda se repite"><font color='green'>longitudes de onda</font></span>.

<div class='alert alert-info'>
    <font size='3'><b>Inspecciona el espectro de abajo.</b></font>
</div>


```python
lightcurve_sliderES.example_spectra(atmospheres=['CO2'])
```

#### Sin atmósfera

Si un planeta no tiene atmósfera, todas las <span title="En una onda periódica la longitud de onda es la distancia física entre dos puntos a partir de los cuales la onda se repite"><font color='green'>longitudes de onda</font></span> estarán en el límite rocoso del planeta. Por lo tanto, un planeta sin aire se verá del mismo tamaño en todas las longitudes de onda.

    <font size='3'><b>Inspecciona el espectro de abajo.</b></font>
